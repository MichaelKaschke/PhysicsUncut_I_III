```python
import numpy as np
import matplotlib.pyplot as plt

# Axes (match MATLAB: -2π : 0.5 : 2π)
dx = 0.5
n_points = int((4*np.pi)/dx) + 1
x = np.linspace(-2*np.pi, 2*np.pi, n_points)
y = np.linspace(-2*np.pi, 2*np.pi, n_points)

# Grid of sample points
X, Y = np.meshgrid(x, y, indexing="xy")

# Velocity field (u, v)
# Here: u = -sin(y), v = sin(x)
U = -np.sin(Y)
V =  np.sin(X)



# Plot: velocity vectors
plt.figure(figsize=(8, 7))

magnitude = np.hypot(U, V) 

plt.quiver(X, Y, U, V, magnitude, cmap="viridis", angles='xy', scale_units='xy', scale=2, pivot='tail')
plt.colorbar(label="Magnitude", shrink=0.9)
plt.gca().set_aspect('equal', adjustable='box')

# Labels & title
plt.xlabel('x-axis', fontsize=12)
plt.ylabel('y-axis', fontsize=12)
plt.title('Velocity vector plot', fontsize=14)

# Grid & limits (optional)
plt.xlim(1.1*x.min(), 1.1*x.max())
plt.ylim(1.1*y.min(), 1.1*y.max())

plt.tight_layout()
plt.show()
```
