# FlettnerRotoren

The program computes the power of a Flettner rotor

## Forward Force

The forward force of a Flettner rotor is given by

$$
F = 2 \pi \rho_L L R^2 v \omega
$$

where

- $\rho_L$ = air density
- $L$ = rotor height
- $R$ = rotor radius
- $v$ = crosswind velocity
- $\omega$ = angular velocity

```python
import numpy as np
```

## Parameters

```python
L = 30        # Height of the rotor
R = 5         # Radius of the rotor
omega = 3     # Rotational speed of the rotor in 1/s approximately 180 rpm
v = 20        # Crosswind velocity in m/s
rhoL = 1.204  # Air density kg/m^3
```

## Calculate forward force

```python
# Forward force in N
F = 2 * np.pi * rhoL * L * R**2 * v * omega
```

## Power

The propulsion power is

$$
P = F \cdot v_s
$$

where $v_s$ is the ship propulsion velocity.

```python
# Ship propulsion velocity in m/s
vs = 6  

# Power in W
P = F * vs
```

## Output

```python
print(f" Forward force in N : {F:6.2e}")
print(f" Power in kW :        {P/1000:6.2e}")
```
