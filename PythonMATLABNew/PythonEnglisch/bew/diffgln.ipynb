# Differenzengleichung

Lösung der Differenzengleichung 
$$x_{k+1} = a \cdot x_k + b , k=1...n$$
unter Variation der Parameter a,b,x_0 und n

_up. 2025-06-30_

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
# Interaktive Steuerelemente
a_slider = widgets.FloatSlider(value=-0.6, min=-1.5, max=1.5, step=0.1, description='a:')
b_slider = widgets.FloatSlider(value=1.0, min=-2.0, max=2.0, step=0.1, description='b:')
x0_slider = widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='x₀:')
n_dropdown = widgets.Dropdown(options=[20, 50, 100], value=50, description='n:')
```

### Ausgabe-Plot

```python

output = widgets.Output()

def update_plot(change=None):
    with output:
        clear_output(wait=True)
        # Parameter übernehmen
        a = a_slider.value
        b = b_slider.value
        x0 = x0_slider.value
        n = n_dropdown.value

        # Berechnung der Folge
        x = [x0]
        for k in range(n):
            x.append(a * x[k] + b)

        # Plot anzeigen
        plt.figure(figsize=(8, 4))
        plt.plot(x, '.-')
        plt.xlabel('Schritt k')
        plt.title(f'Differenzengleichung: xₖ₊₁ = {a:.2f}·xₖ + {b:.2f}')
        plt.grid(True)
        plt.show()

# Widgets verbinden
for w in [a_slider, b_slider, x0_slider, n_dropdown]:
    w.observe(update_plot, names='value')

# Darstellung
display(widgets.VBox([a_slider, b_slider, x0_slider, n_dropdown]))
display(output)

# Initiales Rendering
update_plot()
```
